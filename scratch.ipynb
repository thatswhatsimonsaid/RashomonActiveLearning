```python
import pickle
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import GradientBoostingClassifier
import warnings

# Suppress convergence warnings for Logistic Regression on small/noisy datasets
warnings.filterwarnings("ignore")

### Path Setup ###
try:
    # Works in .py scripts
    PROJECT_ROOT = Path(__file__).resolve().parent.parent.parent
except NameError:
    # Fallback for Jupyter Notebooks (.ipynb)
    # This assumes you are running the notebook from the project root
    PROJECT_ROOT = Path.cwd()

DATA_DIR = PROJECT_ROOT / "src" / "data"

# List of datasets from your LaTeX table to process
DATASETS_TO_PROCESS = [
    # --- UCI / SORTD Benchmarks (1-20) ---
    "anneal", 
    "bank_marketing", 
    "banknote_authentication", 
    "bar-7", 
    "biodeg", 
    "breast_cancer_wisconsin", 
    "cheap_restaurant", 
    "coffee_house", 
    "expensive_restaurant", 
    "haberman", 
    "hepatitis", 
    "hypothyroid", 
    "lymph", 
    "monk1", 
    "monk2", 
    "monk3", 
    "primary-tumor", 
    "tic-tac-toe", 
    "vote", 
    "yeast",
    
    # --- Synthetic Stress Tests (21-29) ---
    "Synthetic_XOR_Baseline", 
    "Synthetic_XOR_Alpha_25", 
    "Synthetic_XOR_Alpha_50",
    "Synthetic_XOR_Alpha_75", 
    "Synthetic_XOR_Alpha_100", 
    "Synthetic_XOR_Phi_05",
    "Synthetic_XOR_Phi_10", 
    "Synthetic_XOR_Phi_25", 
    "Synthetic_XOR_Phi_45",
    
    # --- Logical Stress Tests (Parity) ---
    "Parity_8bit_Noise_00", 
    "Parity_8bit_Noise_06", 
    "Parity_8bit_Noise_16", 
    "Parity_8bit_Noise_26"
]

def get_baseline_scores(file_key):
    path = DATA_DIR / f"{file_key}.pkl"
    if not path.exists():
        return None

    with open(path, "rb") as f:
        df = pickle.load(f)
    
    X = df.drop(columns="Y")
    y = df["Y"]

    # 1. Linear Baseline (Logistic Regression)
    lr = LogisticRegression(max_iter=1000)
    lr.fit(X, y)
    lin_score = lr.score(X, y) * 100

    # 2. Ensemble Baseline (Gradient Boosting)
    gbm = GradientBoostingClassifier(n_estimators=100, random_state=42)
    gbm.fit(X, y)
    gbm_score = gbm.score(X, y) * 100

    return lin_score, gbm_score

def main():
    print(f"Project Root Detected: {PROJECT_ROOT}")
    print(f"{'Dataset':<30} | {'Lin%':<8} | {'GBM%':<8}")
    print("-" * 52)

    for ds in DATASETS_TO_PROCESS:
        scores = get_baseline_scores(ds)
        if scores:
            lin, gbm = scores
            print(f"{ds:<30} | {lin:>7.1f}% | {gbm:>7.1f}%")
        else:
            print(f"{ds:<30} | File not found at {DATA_DIR}")

if __name__ == "__main__":
    main()
```
